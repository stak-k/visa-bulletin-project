```python
import pdfplumber
from pathlib import Path
```

```python
test_pdfs = {
    "Format A - 2016": Path("../data/raw_pdfs/2016/october_2016.pdf"),
    "Accessible - 2023": Path("../data/raw_pdfs/2023/october_2023.pdf"),
    "Format B - 2025": Path("../data/raw_pdfs/2025/november_2025.pdf")
}

for name, path in test_pdfs.items():
    print(name, "->", path.exists())
```

```python
for name, path in test_pdfs.items():
    print("\n" + "=" * 60)
    print(name)

    with pdfplumber.open(path) as pdf:
        print("Page count:", len(pdf.pages))

        text = "\n".join(
            page.extract_text() or ""
            for page in pdf.pages
        )

        print(text[:2000])
```

```python
for name, path in test_pdfs.items():
    print("\n" + "=" * 60)
    print(name)

    with pdfplumber.open(path) as pdf:
        text = "\n".join(
            page.extract_text() or ""
            for page in pdf.pages
        )

    position = text.find("FAMILY-SPONSORED")

    if position != -1:
        print(text[position:position + 2500])
    else:
        print("FAMILY-SPONSORED section not found")
```

```python
for name, path in test_pdfs.items():
    print("\n" + "=" * 60)
    print(name)

    with pdfplumber.open(path) as pdf:
        text = "\n".join(
            page.extract_text() or ""
            for page in pdf.pages
        )

    for line in text.splitlines():
        if "family" in line.lower():
            print(line)
```

```python
path = test_pdfs["Format B - 2025"]

with pdfplumber.open(path) as pdf:
    text = "\n".join(
        page.extract_text() or ""
        for page in pdf.pages
    )

for line in text.splitlines():
    if (
        "family" in line.lower()
        or "final action" in line.lower()
        or "dates for filing" in line.lower()
    ):
        print(line)
```

```python
for name, path in test_pdfs.items():
    print("\n" + "=" * 60)
    print(name)

    with pdfplumber.open(path) as pdf:
        for page_number, page in enumerate(pdf.pages, start=1):
            text = (page.extract_text() or "").lower()

            if "final action dates" in text and "family-sponsored" in text:
                print("Possible family table on page:", page_number)

                tables = page.extract_tables()

                print("Tables found:", len(tables))

                for table in tables:
                    for row in table[:8]:
                        print(row)
```

```python
path = test_pdfs["Format A - 2016"]

with pdfplumber.open(path) as pdf:
    for page_number, page in enumerate(pdf.pages, start=1):
        text = page.extract_text() or ""

        if "F1" in text and "F2A" in text:
            print("Page:", page_number)
            print(text)
```

```python
import re

path = test_pdfs["Format A - 2016"]

with pdfplumber.open(path) as pdf:
    page = pdf.pages[1]   # page 2
    text = page.extract_text() or ""

for line in text.splitlines():
    if re.match(r"^(F1|F2A|F2B|F3|F4)\b", line):
        print(line)
```

```python
path = test_pdfs["Format A - 2016"]

with pdfplumber.open(path) as pdf:
    text = "\n".join(
        page.extract_text() or ""
        for page in pdf.pages
    )

upper_text = text.upper()

final_start = upper_text.find(
    "A. FINAL ACTION DATES FOR FAMILY-SPONSORED"
)

filing_start = upper_text.find(
    "B. DATES FOR FILING FAMILY-SPONSORED"
)

final_section = text[final_start:filing_start]
filing_section = text[filing_start:]
```

```python
pattern = r"^(F1|F2A|F2B|F3|F4)\s+(?:\d{2}[A-Z]{3}\d{2}|C|U)(?:\s+(?:\d{2}[A-Z]{3}\d{2}|C|U)){4}\s*$"

print("FINAL ACTION DATES")

for line in final_section.splitlines():
    if re.match(pattern, line):
        print(line)

print("\nDATES FOR FILING")

for line in filing_section.splitlines():
    if re.match(pattern, line):
        print(line)
```
